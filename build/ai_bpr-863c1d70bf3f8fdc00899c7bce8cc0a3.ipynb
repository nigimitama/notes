# AI BPR (AI-driven Business Process Re-Engineering)

**AI BPR (AI-driven Business Process Re-Engineering)** はAIエージェントによる業務変革手法。  
BPRは業務プロセスを **作り直す** こと。「既存の業務プロセスの一部をAIに置き換える」のではなく、「AI時代に合わせて業務プロセスを作り直す」というアプローチ。
