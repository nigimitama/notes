# よさそうな文献・サイト

## 知識グラフ・依存関係の可視化

#### [理系数学ネットワーク](https://mathnet.penbato.com/)

問題と、それを解くのに使う知識（公式・解法）を3Dネットワークで結んだサイト。知識をクリックすると公式と使いどころ、その知識を使う問題の一覧が出る。層・解法の種類・章・レベル・分野で絞り込める

#### [Math Graph (mckoss)](https://mckoss.com/math-graph/)

数え上げから学部数学までの概念の依存関係を有向グラフで可視化したインタラクティブなサイト。ソースは [GitHub](https://github.com/mckoss/math-graph)

#### Lean Blueprint

Lean（定理証明支援系）での形式化プロジェクトで使われる、定義・補題・定理の依存グラフを生成するツール

- [leanblueprint](https://github.com/PatrickMassot/leanblueprint)
- 例：[PFR予想の形式化の依存グラフ](https://teorth.github.io/pfr/blueprint/dep_graph_document.html)（Terence Taoら）

#### [Mathematics in Mathlib](https://leanprover-community.github.io/mathlib-overview.html)

Leanの数学ライブラリMathlibで形式化済みの数学の分野・定理の一覧

## 百科事典・Wiki

#### [ProofWiki](https://proofwiki.org/wiki/Main_Page)

定義・定理とその証明を集めたWiki

#### [Wolfram MathWorld](https://mathworld.wolfram.com/)

Wolframによる数学百科事典

#### [MaRDI Portal](https://portal.mardi4nfdi.de/)

ドイツの数学研究データ基盤（MaRDI）による、論文・数式・ソフトウェア・データセットなどを結んだ数学の知識グラフ（Wikibaseベース）

#### [MSC2020 (zbMATH)](https://zbmath.org/classification/)

数学分野の国際的な分類体系（Mathematics Subject Classification）。分野の全体像を把握するのに便利

## 日本語のサイト

#### [数学の景色](https://mathlandscape.com/)

大学数学の定義・定理を証明付きで解説するサイト

#### [高校数学の美しい物語](https://manabitimes.jp/math)

高校〜大学初年級の数学のトピックを解説するサイト
