# トークノミクス

## トークノミクスとは

**トークノミクス（tokenomics）** は token と economics を合わせた造語で、トークンの

- **供給**：総量、発行スケジュール、バーン（焼却）
- **配分**：誰に、どれだけ、いつ渡すか（ベスティング）
- **需要と価値の源泉**：なぜ人はそのトークンを持ちたいのか
- **インセンティブ**：保有・利用・ステーキングなどの行動に対して何を報酬とするか

の設計を指す。

BitcoinやEthereumのネイティブトークンは、マイナーやバリデーターに正直に振る舞わせるための報酬として設計されている（→[コンセンサス](../blockchain/consensus.ipynb)、[Proof of Stake](../ethereum/proof_of_stake.ipynb)）。これらのプロトコルが持つ「トークン報酬で見知らぬ参加者を協力させる」仕組みを、DAppsでも再現しようとするのがトークノミクスの出発点である。

用語としては2017年頃から学術的にも使われるようになり、当初はトークンの用途やICOによる資金調達など「トークンの価値の設計」を指していたが、次第にコンセンサスの経済学（[クリプトエコノミクス](./cryptoeconomics.ipynb)）やゲーム理論的なインセンティブ全般も含む広い意味で使われるようになった（Mougayar, 2017; Ennis et al., 2018; Kampakis, 2022）。本ページでは狭い意味、つまり **自律性のためのトークン価値の設計** を扱う（→[エコノミクス](./index.ipynb)）。

ただし実態としては、トークンの発行が資金調達やマーケティングの手段になっている面も大きい。設計がまずいと、価格の下落 → 報酬の価値の低下 → 参加者の離脱 → さらなる下落、という負のスパイラルに陥る。

## トークンの分類

トークンは機能によって大まかに次のように分けられる（技術的な規格は [トークン規格](../ethereum/token_standards.ipynb) を参照）。

| 種類 | 役割 | 例 |
| --- | --- | --- |
| ネイティブトークン | L1/L2の手数料（gas）の支払い、ステーキングの担保 | BTC, ETH, SOL |
| ガバナンストークン | プロトコルの仕様やパラメーター、トレジャリーの使途を決める投票権（→[DAO](../applications/dao.ipynb)） | UNI, AAVE, MKR / SKY |
| ユーティリティトークン | サービス内での支払い、利用権、割引 | FIL（ストレージの対価）, LINK（オラクルの対価） |
| ステーブルコイン | 法定通貨などと価値を連動させる（→[ステーブルコイン](../applications/defi/stablecoins.ipynb)） | USDC, USDT, DAI |
| 持ち分トークン | プールに預けた資産の持ち分を表す | UniswapのLPトークン, AaveのaToken, Lidoの stETH |
| NFT | 唯一性のある資産や会員権（→[NFT](../applications/nft.ipynb)） | Nouns, ENS |
| セキュリティトークン | 株式・社債・不動産などの権利をトークン化したもの。証券として規制される | RWA、日本のデジタル証券 |

1つのトークンが複数の役割を兼ねることも多い。例えばETHは手数料の支払い手段であり、ステーキングの担保であり、DeFiの担保資産でもある。

## 市場価値をどう担保するか

トークンがインセンティブとして機能するには、まず市場で価格が付く必要がある。ミクロ経済学では、価格は

- **供給側**：追加で1単位を供給するのにかかるコスト（**限界費用**）
- **需要側**：追加で1単位を得たときに増える満足度（**限界効用**）

が釣り合うところで決まる。売り手は限界費用より高く売れなければ供給せず、買い手は限界効用より安く買えるなら買い続ける。トークノミクスの設計は、この両側をどう作り込むかと整理できる。

### 供給側：限界費用

| 方式 | 新しいトークン1単位を得るための限界費用 | 補足 |
| --- | --- | --- |
| Bitcoin（PoW） | マイニングに費やす計算資源（電気代、ハードウェア） | 難易度調整で安定する。半減期のたびに1 BTCあたりのコストは約2倍になる |
| Ethereum（PoS） | ステーキングの **機会費用**（32 ETHをロックせずに運用していれば得られたリターン） | stETHなどのリキッドステーキングはロック中も運用できるため、機会費用を下げ、結果としてETHの価値を支える力を弱める可能性がある |
| ICO | 販売トークン1単位の購入に必要な他のトークン（多くはETH） | |
| ロックドロップ（lockdrop） | 一定期間ロックした他のトークンの機会費用 | ロック期間が長いほど多くのトークンがもらえる |
| プレマイン（全量を事前発行） | 実質的にゼロ | 配布を受けた側にはコストがかかっていない |

経済学の研究は、設計よりも価値評価（valuation）が中心である。

- **Bitcoin**：マイニングのコストは定量的に把握できるので、それを価格の下限・基準とみなす **生産コストモデル**（Hayes, 2017, 2019）
- **Ethereum**：PoSの機会費用を「金融市場で同程度のリスクに対して期待されるリターン」で推定し、価値を評価する（Fanti et al., 2019）

### 需要側：限界効用

| 類型 | 限界効用の源泉 | 例 |
| --- | --- | --- |
| 通貨 | P2Pの電子現金としての利用、手数料の支払い。利用者が増えるほど価値が高まる（ネットワーク効果） | BTC（デジタルゴールド） |
| 燃料 | DAppsを動かすためのgas | ETH（デジタルオイル） |
| 株式のアナロジー | ガバナンスへの参加、利益の分配 | ガバナンストークン、LPトークン |
| アートのアナロジー | デザイン、レアリティ、コミュニティへの所属 | CryptoPunks, CryptoKitties |
| 両者の組み合わせ | アートであり、かつガバナンストークン | Nouns DAO（→[DAO](../applications/dao.ipynb)） |

限界効用の定量化の方法としては、次のような研究がある。

- **Bitcoin**：アドレス数やノード数から推定した送金ネットワークの規模を代理指標とする
- **一般的なトークン**：現在のユーザー数とトランザクション数から価値を推定する動学モデル（Cong, Li & Wang, 2021）や、流通速度・ステーキング比率・供給の増加率から推定するモデル（Liu, 2022）
- **定期的に収益が発生するトークン**（LPトークンなど）：株式と同じDCF法を応用する
- **NFT**：保有アドレス数や取引数に加えて、画像や説明文の特徴を使う

:::{admonition} コラム：Bitcoinの価値の源泉はPoWの電気代か
:class: note

PoWとPoSの比較では「PoWのBitcoinは電気代という裏付けがあるが、PoSに移行したETHには裏付けがない」という主張がよく見られる。しかしこれは、19世紀の経済学における **費用価値説** と **効用価値説** の論争の繰り返しである。

この論争にはMarshall（1890）が決着を付けている。

> 価値の源泉が費用なのか効用なのかを論じるのは、紙を切っているのがハサミの上の刃なのか下の刃なのかを論じるようなものである。

価格は限界費用と限界効用の両方で決まるので、どちらか一方だけを価値の源泉とみなす議論は成り立たない。
:::

## 価格をどう安定させるか

価格が付くだけでなく、その価格が激しく変動しないことも、インセンティブとして機能するためには重要である。報酬の価値が予測できなければ、マイナーもバリデーターも参加の判断ができない。

### プロトコル層のスタビライザー

| 仕組み | 本来の目的 | 価格の安定化の観点から見た役割 |
| --- | --- | --- |
| 難易度調整（Bitcoin） | ブロック間隔を約10分に保つ | 限界費用の急な変動を抑え、予測しやすくする |
| 手数料のオークション（Bitcoin, Ethereum） | ネットワークの混雑を平準化する | 限界効用の急な変動を抑え、予測しやすくする |

Bitcoinは2,016ブロックごとに難易度を調整するが、ハッシュレートが急変すると調整が追いつかない。これに対しては、ブロックごとに調整するほうが安定する（Noda, Okumura & Hashimoto, 2020）、逆にブロック間隔がしきい値を超えたときだけ調整すべき（Saito & Iwamura, 2019）といった分析がある。

手数料の決め方（**Transaction Fee Mechanism, TFM**）はオークション理論の応用として研究されている。

| 方式 | 内容 | 課題 |
| --- | --- | --- |
| 第一価格オークション | 各送信者が自分の入札額をそのまま支払う。Bitcoinと、EIP-1559以前のEthereum | 正直な入札が最適でなく、入札額の推測ゲームになる |
| 第二価格オークション | 落札者は「次に高い入札額」を支払う | ブロック生成者が偽のトランザクションを入れて価格をつり上げられる |
| 単一価格オークション（monopolistic price） | 同じブロック内のトランザクションから同じ金額を徴収する（Lavi, Sattath & Zohar, 2019） | ブロックが大きいと近似的に誘因両立的になる |
| EIP-1559 | プロトコルが決めるベースフィー（バーン）＋送信者が決めるチップ（priority fee）の2階建て（→後述の「バーン」） | 多くの状況で正直な入札が最適になる（Roughgarden, 2021） |

一方で、利用者・ブロック生成者・両者の談合のすべてに対して同時に誘因両立的なTFMは作れないという不可能性定理も示されている（Chung & Shi, 2023）。

### ペッグ

より直接的な安定化の方法は、トークンと別の資産との交換比率をあらかじめ決めておくこと。

- **ステーブルコイン**：交換比率そのものを固定する（→[ステーブルコイン](../applications/defi/stablecoins.ipynb)）
  - 資産担保型（USDT, PAX Gold）：ペッグ対象の資産を発行主体が保有する
  - 暗号資産担保型（DAI）：誰でも担保を預けて発行できるので分散的だが、過剰担保が必要
  - アルゴリズム型（Basis, TerraUSD）：担保なしでペッグを保とうとするが、実用に耐えたものはまだない
- **交換比率のルールを固定する**：交換比率そのものではなく、交換比率の決め方をあらかじめコードで決める
  - **トークンボンディングカーブ（TBC）**：ETHを預けるとトークンが発行され、トークンを返すとETHが戻る。交換比率はトークンの供給量の関数として決まる
  - **AMM**：TBCを2種類のトークンに拡張したもの。交換比率はプール内の2つのトークンの量で決まる（→[AMM](../applications/defi/amm.ipynb)）

TBCの例として、価格が供給量 $S$ に比例する線形のカーブ $p(S) = aS$ を考える。供給量を $S_0$ から $S_1$ に増やすのに必要なETHは、カーブの下の面積

$$
\int_{S_0}^{S_1} aS \, dS = \frac{a}{2}\left(S_1^2 - S_0^2\right)
$$

で、売却時も同じカーブに沿ってETHが払い戻される。預けられたETHは常にカーブの下の面積に等しいので、いつでも売却に応じられる。

```python
import matplotlib.pyplot as plt
import japanize_matplotlib  # noqa: F401
import numpy as np


class LinearBondingCurve:
    # 価格 p(S) = a * S の線形ボンディングカーブ

    def __init__(self, a: float):
        self.a = a
        self.supply = 0.0
        self.reserve = 0.0  # 預けられたETH

    def price(self) -> float:
        return self.a * self.supply

    def buy(self, amount: float) -> float:
        s0, s1 = self.supply, self.supply + amount
        cost = self.a / 2 * (s1**2 - s0**2)
        self.supply, self.reserve = s1, self.reserve + cost
        return cost

    def sell(self, amount: float) -> float:
        s0, s1 = self.supply, self.supply - amount
        refund = self.a / 2 * (s0**2 - s1**2)
        self.supply, self.reserve = s1, self.reserve - refund
        return refund


tbc = LinearBondingCurve(a=0.001)
alice_cost = tbc.buy(100)  # 早期の購入者
bob_cost = tbc.buy(400)  # 後から大きく買う
alice_refund = tbc.sell(100)
print(f"Alice: {alice_cost:.1f} ETHで100枚購入 → Bobの購入後に売却して {alice_refund:.1f} ETH（{alice_refund / alice_cost:.0f}倍）")
bob_value = tbc.a / 2 * tbc.supply**2  # いま全量を売ったときの払い戻し額
print(f"Bob: {bob_cost:.1f} ETHで400枚購入 → Aliceの売却後に全量を売ると {bob_value:.1f} ETH")
print(f"リザーブ {tbc.reserve:.1f} ETH = 供給量{tbc.supply:.0f}枚のカーブの下の面積 {0.001 / 2 * tbc.supply**2:.1f} ETH")

s = np.linspace(0, 600, 300)
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(s, 0.001 * s, label="価格 $p(S)=aS$")
ax.fill_between(s[s <= 100], 0.001 * s[s <= 100], alpha=0.4, label="Aliceの購入（0→100枚）")
mask = (s >= 100) & (s <= 500)
ax.fill_between(s[mask], 0.001 * s[mask], alpha=0.4, label="Bobの購入（100→500枚）")
ax.set(xlabel="供給量 $S$", ylabel="価格 [ETH]", title="線形トークンボンディングカーブ")
ax.legend()
plt.show()
```

価格がカーブで決まるので、取引相手を探さずにいつでも売買でき、価格（ETH建て）が自律的に決まる。一方で、カーブが右上がりなら **早く買った人ほど有利** になり、後から買った人の資金で早期の購入者が利益を得る構造でもある。アーティストのNFTの販売などへの応用が考えられている。

TBCやAMMの経済学的な研究は定式化が中心で、新しいモデルの提案は少なく、設計は実務家（DEXの開発組織など）が先行している。しかしTBCやAMMは予測市場のマーケットメイカー（Hanson, 2003; Pennock, 2004）の応用であり（→[予測市場](../applications/prediction_markets.ipynb)）、CPAMMの $xy = k$ は経済学の無差別曲線と同じ形をしているので、経済学が貢献できる余地は大きいと考えられる。

ステーブルコインについては、資産担保型・暗号資産担保型には通貨危機や銀行の取り付け騒ぎのモデル、アルゴリズム型にはポンジスキームのモデルを応用して、頑健性や失敗を分析する研究がある。

## 供給の設計

### 最大供給量と発行スケジュール

| 方針 | 内容 | 例 |
| --- | --- | --- |
| 固定供給 | 上限を決め、発行ペースを徐々に下げる | Bitcoin（上限約2,100万BTC） |
| 継続的なインフレ | 上限を決めず、セキュリティの報酬として毎年一定程度を発行し続ける | ETH, SOL, DOT |
| 一括発行 | 最初に全量を発行し、配布はベスティングで時間をかけて行う | UNI（10億枚）, 多くのガバナンストークン |
| 需要に応じて伸縮 | 担保や償還に応じて発行・バーンする | ステーブルコイン |

Bitcoinは1ブロックあたりの報酬が50 BTCから始まり、210,000ブロック（約4年）ごとに半分になる **半減期（halving）** を持つ。報酬の等比級数の和が上限となる。

$$
\sum_{i=0}^{\infty} 210{,}000 \times \frac{50}{2^i} = 210{,}000 \times 100 = 21{,}000{,}000
$$

実際には報酬はsatoshi（$10^{-8}$ BTC）単位で切り捨てられるため、上限はこれをわずかに下回る。

```python
BLOCKS_PER_HALVING = 210_000
BLOCKS_PER_YEAR = 6 * 24 * 365  # 10分に1ブロック

heights = np.arange(0, BLOCKS_PER_HALVING * 33, 1_000)
era = heights // BLOCKS_PER_HALVING
reward = 50 / 2.0**era
supply = np.cumsum(reward * 1_000)
years = 2009 + heights / BLOCKS_PER_YEAR
inflation = reward * BLOCKS_PER_YEAR / supply * 100

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
axes[0].plot(years, supply / 1e6)
axes[0].axhline(21, color="gray", linestyle="--", linewidth=1)
axes[0].set(xlim=(2009, 2060), xlabel="年", ylabel="供給量 [百万BTC]", title="Bitcoinの供給量")
axes[1].plot(years, inflation)
axes[1].set(xlim=(2012, 2060), ylim=(0, 15), xlabel="年", ylabel="年率 [%]", title="供給量の増加率（インフレ率）")
plt.tight_layout()
plt.show()

print(f"2024年の半減期後の報酬: {50 / 2**4} BTC/ブロック")
print(f"2140年頃までの総供給量: {supply[-1]:,.0f} BTC")
```

供給の増加率は半減期ごとに半分になり、2024年の4回目の半減期の後は年率1%を下回った。

固定供給のトークンは「希少性」を訴求しやすい一方で、ブロック報酬がゼロに近づくと、マイナーへの報酬を手数料だけでまかなえるのかという **セキュリティ予算（security budget）** の問題が生じる。継続的なインフレを採用するチェーンは、この問題を避ける代わりに、保有者の持ち分が希薄化する。

### バーン

**バーン（burn）** は、誰も秘密鍵を知らないアドレスに送るなどしてトークンを永久に使えなくすること。供給を減らす手段として使われる。

Ethereumでは2021年の **EIP-1559** で、手数料のうちベースフィー（base fee）をバーンする仕組みが導入された。PoSのステーキング報酬による新規発行と、ベースフィーのバーンとの差し引きでETHの供給量が決まる。

$$
\Delta S = \underbrace{\text{発行量}}_{\text{ステーキング報酬}} - \underbrace{\text{バーン量}}_{\text{ベースフィー} \times \text{gas使用量}}
$$

ネットワークが混雑してバーン量が発行量を上回ると、ETHの供給は減少する（デフレ）。The Merge（2022年）後の一時期はこの状態になり、「ultrasound money」と呼ばれた。しかし2024年のDencun（EIP-4844）でL2のデータ掲載コストが大幅に下がり、L1のベースフィーも低くなったため、その後はバーン量が発行量を下回って供給が増加に転じている。

PoSの新規発行量は、ステーキングされているETHの総量 $D$ の平方根にほぼ比例する（バリデーター1人あたりの利回りは $1/\sqrt{D}$ に比例する）。年間の発行量はおおよそ

$$
\text{年間発行量} \approx 166 \sqrt{D} \ \text{ETH}
$$

で、$D = 3{,}400$万 ETH なら年間約97万ETH（総供給量の約0.8%）になる。

### 流通供給量とFDV

トークンの規模を見るときは、次の2つを区別する必要がある。

| 指標 | 計算 | 意味 |
| --- | --- | --- |
| 時価総額（Market Cap） | 価格 × 流通供給量 | いま市場で売買できるトークンの価値 |
| 完全希薄化後評価額（FDV, Fully Diluted Valuation） | 価格 × 最大供給量（または総供給量） | 将来すべてのトークンが流通した場合の価値 |

新しいトークンの多くは、上場時に総供給量の10〜20%しか流通していない（**ローフロート**）。流通量が少ないため少額の買いで価格が上がり、FDVが非常に高くなる。その後、チームや投資家のトークンのロックが解除される（**アンロック**）たびに売り圧力がかかり、価格が下がりやすい。「低流通・高FDV」はトークンの価格が長期的に低迷する典型的なパターンとされる。

## 配分とベスティング

### 典型的な配分

| 配分先 | 割合の目安 | 目的 |
| --- | --- | --- |
| コミュニティ（エアドロップ、流動性マイニング、エコシステム基金） | 40〜60% | 利用者の獲得、分散化 |
| チーム・アドバイザー | 15〜25% | 開発者への報酬 |
| 投資家（VC） | 15〜25% | 開発資金の調達 |
| トレジャリー・財団 | 10〜20% | 将来の開発や助成 |

例えばUniswapのUNIは、総量10億枚のうち60%をコミュニティ（うち15%は過去の利用者へのエアドロップ）、21.3%をチーム、18%を投資家、0.7%をアドバイザーに配分した。

### ベスティング

チームや投資家のトークンには、一定期間売却できない **ロックアップ** と、少しずつ受け取れるようにする **ベスティング（vesting）** が設定されるのが普通。

- **クリフ（cliff）**：最初の一定期間（例：1年）は何も受け取れず、期間が過ぎた時点でまとめて受け取る
- **リニアベスティング**：クリフ後、残りを毎月・毎ブロック均等に受け取る

「4年ベスティング・1年クリフ」（1年後に1/4をまとめて受け取り、残りを3年かけて均等に受け取る）などが一般的。スタートアップのストックオプションと同じ発想で、開発者が早期に売り抜けるのを防ぎ、長期的な貢献を促す。

```python
months = np.arange(0, 61)


def vesting(months: np.ndarray, share: float, cliff: int, duration: int, tge: float = 0.0) -> np.ndarray:
    # TGE（Token Generation Event）時に tge の割合を解放し、残りを duration か月かけて線形に解放する。
    # クリフまでは何も解放せず、クリフの時点でそれまでの分をまとめて解放する
    linear = np.clip(months / duration, 0, 1) * (1 - tge)
    linear = np.where(months >= cliff, linear, 0)
    return share * (tge + linear)


schedule = {
    "コミュニティ": vesting(months, 0.45, cliff=0, duration=48, tge=0.15),
    "トレジャリー": vesting(months, 0.15, cliff=0, duration=48),
    "投資家": vesting(months, 0.20, cliff=12, duration=36),
    "チーム": vesting(months, 0.20, cliff=12, duration=48),
}

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
axes[0].stackplot(months, *[v * 100 for v in schedule.values()], labels=schedule.keys())
axes[0].set(xlabel="TGEからの月数", ylabel="流通供給量 [総供給量に対する%]", title="アンロックのスケジュール（例）")
axes[0].legend(loc="upper left", fontsize=8)

circulating = sum(schedule.values())
monthly_unlock = np.diff(circulating, prepend=0) * 100
axes[1].bar(months[1:], monthly_unlock[1:])
axes[1].set(xlabel="TGEからの月数", ylabel="その月のアンロック量 [%]", title="月ごとのアンロック量")
plt.tight_layout()
plt.show()

print(f"TGE時の流通割合: {circulating[0]:.1%}")
print(f"11か月目の流通割合: {circulating[11]:.1%} → 12か月目（クリフ明け）: {circulating[12]:.1%}")
```

TGE（Token Generation Event、トークンの発行・上場時点）の流通割合は1割未満で、クリフ明けにアンロック量が急増する。アンロックの日程は事前に公開されているため市場はある程度織り込むが、実際にはアンロックの前後に価格が下がる傾向が観察されている。

## 配布の方法

| 方法 | 内容 | 課題 |
| --- | --- | --- |
| ICO（Initial Coin Offering） | プロジェクトが直接トークンを販売する。2017年に大流行 | 詐欺や実体のないプロジェクトが多発し、多くの国で規制対象に |
| IEO（Initial Exchange Offering） | 暗号資産取引所が審査・販売を代行する | 取引所への信頼に依存 |
| IDO（Initial DEX Offering） | DEXやローンチパッドで販売する | ボットによる買い占め |
| エアドロップ | 過去の利用者などに無償で配る | 報酬目当ての大量アカウント（シビル） |
| 流動性マイニング | 流動性の提供や借り入れの量に応じてトークンを配る | 報酬目当ての資金はすぐに去る |
| ポイントプログラム | トークン発行前に「ポイント」を配り、後でトークンに交換する | 交換比率が不透明 |

### 流動性マイニング

2020年6月、レンディングプロトコルのCompoundが、利用者に毎日COMPトークンを配り始めた（→[レンディング](../applications/defi/lending.ipynb)）。COMPの価値が高かったため、借入金利を払ってもCOMPの報酬のほうが大きく、「借りれば借りるほど儲かる」状態になった。これが他のプロトコルにも広がり、**DeFi Summer** と呼ばれるブームを引き起こした。

流動性マイニングは利用者を急速に集められるが、報酬の利回りだけを求める資金（**マーセナリーキャピタル**、傭兵資本）は、より利回りの高いプロトコルが現れるとすぐに移動する。報酬として配られたトークンは即座に売られることが多く、価格の下落 → 利回りの低下 → 資金の流出というサイクルに陥りやすい。

### エアドロップとシビル攻撃

エアドロップは過去の利用者に報いる手段として定着したが、エアドロップを期待して1人で大量のアドレスを使い分ける **シビル攻撃** も一般化した。プロジェクト側は、アドレス間の資金の流れの分析、利用期間や利用額の重み付け、本人確認などで対策しているが、完全に防ぐことは難しい。

## 価値の源泉

トークンの価格は、最終的には「誰がなぜそのトークンを買って持ち続けるのか」で決まる。プロトコルが生み出した価値がトークン保有者に還元される仕組みは **バリューアクルーアル（value accrual）** と呼ばれる。

| 仕組み | 内容 | 例 |
| --- | --- | --- |
| 手数料の支払い手段 | プロトコルを使うためにトークンが必要 | ETHのgas、FILのストレージ料金 |
| 手数料の分配 | プロトコルの収益をステーカーに配る | GMX, Curve（veCRVへの手数料分配） |
| バイバック＆バーン | 収益でトークンを買い戻してバーンする | MKR / SKY, BNB, Hyperliquid（HYPE） |
| ステーキング・担保 | ネットワークのセキュリティや保険のためにロックする | ETH, Aaveのセーフティモジュール |
| ガバナンス | 投票権そのもの。収益に影響する決定ができる | UNI, ガバナンストークン全般 |

ガバナンストークンの多くは、長らく投票権以外の経済的な権利を持たなかった。例えばUniswapは、取引手数料の一部をプロトコル側（ひいてはUNI保有者）に振り向ける「フィースイッチ」を実装しながら、規制上のリスク（証券とみなされる懸念）などから長年オンにしてこなかった。株式でいえば「配当も残余財産の分配もない議決権株」に近く、価値の根拠が弱いという批判がある。米国で暗号資産への規制方針が変わりつつあることもあり、2025年以降は手数料をバイバック＆バーンに回す仕組みを導入するプロトコルが増えている。

### ve（vote-escrow）モデル

**veトークノミクス** は、Curve Financeが導入した、トークンを長期間ロックするほど大きな権利が得られる仕組み。

- CRVを最大4年間ロックすると **veCRV** が得られる。veCRVの量はロック期間に比例し（4年なら1 CRV → 1 veCRV、1年なら0.25 veCRV）、満期に向けて線形に減少する
- veCRVの保有者は、プロトコルの取引手数料の分配を受け、自分の流動性提供の報酬が最大2.5倍になり、どのプールにCRVの新規発行（報酬）を多く配分するかを投票（**ゲージ投票**）で決められる
- veCRVは譲渡できない

ゲージ投票で自分のプールに報酬を誘導できれば、そのプールに流動性が集まる。そのため、ステーブルコインの発行体などがveCRVの投票権を奪い合う **Curve Wars** が起きた。Convex Financeは利用者のCRVをまとめて永久ロックし、その投票権を集約する代わりに、譲渡可能なcvxCRVを発行した。さらに投票権を持つ人に「このプールに投票してくれたら報酬を払う」という **賄賂（bribe）** の市場（Votiumなど）まで生まれた。

ロックによって売り圧力を抑え、長期保有者に権利を集中させる設計だが、結果として投票権の売買市場ができ、ガバナンスが資金力で左右されることにもなった。

## トークンの価値評価

### 交換方程式によるアプローチ

ユーティリティトークンの価値評価には、貨幣数量説の **交換方程式** を応用する方法がある（Burniske, 2017）。

$$
MV = PQ
$$

| 記号 | 意味 |
| --- | --- |
| $M$ | トークン経済の規模（トークンの総価値、ドル建て） |
| $V$ | 流通速度（1年間に1枚のトークンが平均何回使われるか） |
| $P$ | サービスの価格（ドル建て） |
| $Q$ | サービスの量 |

$PQ$ はトークンを使って行われる経済活動の総額（GDPに相当）なので、トークンの総価値は

$$
M = \frac{PQ}{V}, \quad \text{トークン1枚の価格} = \frac{M}{\text{流通供給量}}
$$

と見積もれる。

```python
def token_price(annual_gdp: float, velocity: float, circulating_supply: float) -> float:
    # 交換方程式 MV = PQ からトークン1枚の価格を求める
    return annual_gdp / velocity / circulating_supply


gdp = 1_000_000_000  # 年間10億ドル分のサービスがトークンで決済される
supply = 100_000_000  # 流通供給量1億枚
for v in [2, 5, 20, 100]:
    print(f"流通速度 V={v:>3}: 1枚あたり {token_price(gdp, v, supply):6.2f} ドル")
```

### 流通速度の問題

上の計算から分かるとおり、同じ経済規模でも流通速度 $V$ が大きいほどトークンの価値は小さくなる。

ここで問題になるのが、**支払いに使うだけのトークンは保有するインセンティブがない** こと。利用者はサービスを使う直前にトークンを買い、サービス提供者は受け取ったらすぐに法定通貨やステーブルコインに替えればよい。DEXやブリッジで瞬時に交換できるなら、$V$ は際限なく大きくなり、トークンの価値はゼロに近づく。これは **流通速度の問題（velocity problem）** と呼ばれる（Samani, 2017; Buterin, 2017）。

そのため、トークンの価値を保つ設計の多くは、何らかの形で **トークンを保有し続ける理由** を作り、流通速度を下げることを狙っている。

- ステーキングやロック（veモデル）で一定期間動かせなくする
- 保有量に応じた手数料の分配や割引
- ガバナンスの権利
- 担保としての利用

### その他の指標

| 指標 | 計算 | 意味 |
| --- | --- | --- |
| P/F比（Price-to-Fees） | 時価総額（またはFDV） ÷ 年間手数料収入 | 株式のPSRに近い。利用者が払っている手数料の総額に対する評価 |
| P/E比（Price-to-Earnings） | 時価総額 ÷ トークン保有者に還元される年間収益 | 株式のPERに相当。手数料のうち保有者に還元される分だけを見る |
| TVL比 | 時価総額 ÷ TVL（プロトコルに預けられた資産の総額） | DeFiプロトコルの規模に対する評価 |
| 実質利回り（Real Yield） | 手数料収入による利回り | 新規発行による報酬（希薄化）を除いた、本当の収益による利回り |

ステーキング報酬が年率20%でも、それが新規発行のトークンで払われているなら、ステーキングしない保有者から価値が移転しているだけで、全体としての価値は増えていない。報酬が **実際の手数料収入から払われているか、新規発行から払われているか** を区別することが重要である。

## 失敗の典型

### 持続不可能な利回り

高い利回りで資金を集め、その利回りを新規発行のトークンや新規参加者の資金でまかなう設計は、成長が止まると崩壊する。

- **Terra/LUNA（2022年）**：アルゴリズム型ステーブルコインUSTに、レンディングプロトコルAnchorが約20%の固定利回りを提供していた。利回りの原資は実際の運用収益ではまかなえず、補填が続いていた。USTのペッグが外れると、LUNAの大量発行による「死のスパイラル」で、両者合わせて数百億ドルの価値が数日で失われた（→[ステーブルコイン](../applications/defi/stablecoins.ipynb)、[ケーススタディ](./case_studies.ipynb)）
- **OlympusDAO（OHM, 2021年）**：トークンをステーキングすれば年率数千%以上の報酬（新規発行）が得られる設計で、「全員がステーキングし続ければ全員が得をする」というゲーム理論的な主張（(3,3)）を掲げた。新規の買いが止まると価格は最高値から99%以上下落した

### Play-to-Earnのインフレ

**Axie Infinity**（2021年）は、ゲームをプレイすると報酬トークンSLPが得られる「Play-to-Earn」で、フィリピンなどで生計を立てる人が出るほど流行した。しかしSLPの主な使い道は新しいキャラクターの繁殖だけで、新規プレイヤーの流入が止まると発行量が需要を大きく上回り、SLPの価格は暴落した。ゲームを楽しむために払う人がいなければ、稼ぐ人への報酬は新規参加者の資金でまかなうしかない。

### 共通する教訓

- 報酬の原資が **外部からの実際の収益** なのか、**新規発行や新規参加者の資金** なのかを確認する
- 価格の上昇を前提にしたインセンティブは、価格が下がると逆回転する（反射性、reflexivity）
- 流通供給量、アンロックのスケジュール、チームと投資家の配分を確認する

## 規制

トークンが **有価証券（security）** に当たるかどうかは、発行体の規制に大きく影響する。

- **米国**：投資契約かどうかを判定する **Howeyテスト**（①資金の投資、②共同事業、③利益への期待、④他者の努力による利益）が基準とされる。SECは多くのトークンを証券とみなして執行を進めてきたが、2025年以降は暗号資産に関する規制の枠組みづくりが進み、ステーブルコインについてはGENIUS法（2025年）が成立した
- **日本**：決済に使えるトークンは資金決済法上の **暗号資産**、収益の分配を受ける権利をトークン化したものは金融商品取引法上の **電子記録移転権利**（セキュリティトークン）として規制される。暗号資産も投資対象としての性格が強まっていることから、金融商品取引法の対象に移して、インサイダー取引規制などを適用する方向で検討が進められている

「収益を保有者に分配する」設計（バリューアクルーアル）は、トークンの価値の根拠を強める一方で、証券とみなされるリスクも高める。プロトコルがフィースイッチのオンをためらってきた理由の1つはここにある。
