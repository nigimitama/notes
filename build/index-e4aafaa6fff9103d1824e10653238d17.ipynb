# エコノミクス

## ブロックチェーンのエコノミクス

ブロックチェーンには暗号技術と並んで、**経済学的な側面** がある。管理者のいないシステムを動かし続けるには、見知らぬ参加者が自分の利益のために行動した結果として、システム全体が望ましい状態に保たれる必要がある。その仕組みを設計するのがブロックチェーンのエコノミクスで、大きく2つに分けられる。

| | クリプトエコノミクス | トークノミクス |
| --- | --- | --- |
| 問い | 分散性のための合意形成をどう設計するか | 自律性のためのトークン価値をどう設計するか |
| 扱う課題 | 戦略的行動、スパム・シビル攻撃、ただ乗り | 限界費用、限界効用、価格の安定化 |
| 学術的な先例 | アルゴリズミックメカニズムデザイン、分散アルゴリズミックメカニズムデザイン | 新古典派経済学（価値と価格の理論）、金融経済学（金融商品、金融政策、デジタル通貨） |
| ページ | [クリプトエコノミクス](./cryptoeconomics.ipynb) | [トークノミクス](./tokenomics.ipynb) |

これらは「何を作るか（What）」「なぜ作るか（Why）」ではなく「どう作るか（How）」に関わる部分であり、Howのなかでも大工よりは建築家に近い、全体の設計の部分にあたる。

## 用語の歴史

Bitcoin（2009年）の新規性は、P2Pの電子現金における合意形成の問題を、**インセンティブ設計の活用** という、それまでとは違うアプローチで解決したことにある。この新規性はEthereum（2015年）やThe DAO（2016年）などに受け継がれ、それを表す言葉として2014〜2016年頃からクリプトエコノミクスやトークノミクスという言葉が使われ始めた。

- **クリプトエコノミクス**：Ethereumのコミュニティ（Vlad Zamfir, Vitalik Buterin）が2015年頃に使い始め、学術的には「メカニズムデザインの一分野」と位置づけられるようになった
- **トークノミクス**：言葉自体は2012年頃には存在したが、ブロックチェーンの文脈で学術的に使われ始めたのは2017年頃から。当初はトークンの用途やICOによる資金調達の設計を指していたが、次第にクリプトエコノミクスの領域まで含むようになった

## 2つを統合する

2つは互いに依存している。

- 合意形成は、報酬としてのトークンに価値がなければ自律的に続かない（クリプトエコノミクスはトークノミクスを必要とする）
- トークンの価値は、それが合意形成に使われなければ分散性に貢献しない（トークノミクスはクリプトエコノミクスを必要とする）

それぞれには学術的な先例があるので、どちらか一方だけなら必ずしも新しいものではない。**両者を統合してこそ新しい** というのが伊東（2024）の主張である。統合のためには、次の6つの観点を同時に考える必要がある。

```mermaid
flowchart LR
    subgraph ce["クリプトエコノミクス<br/>分散性のための合意形成"]
        s1["戦略的行動"]
        s2["スパム・シビル攻撃"]
        s3["ただ乗り"]
    end
    subgraph te["トークノミクス<br/>自律性のためのトークン価値"]
        t1["限界費用"]
        t2["限界効用"]
        t3["安定化"]
    end
    ce -- "報酬としてのトークンに<br/>価値が必要" --> te
    te -- "トークンが合意形成に<br/>使われる必要" --> ce
```

Bitcoinはこの6つすべてに対処している点で統合のベンチマークになる。一方で、多くのDAppsはどれかが欠けている（→[ケーススタディ](./case_studies.ipynb)）。

## 今後の課題

- **システムの外のインセンティブ**：取引所での空売りと組み合わせた攻撃（ゴールドフィンガー攻撃）や、アプリケーション層のDEXから生まれたMEVのように、プロトコルの外にあるインセンティブが合意形成を壊しうる
- **合理性の仮定**：ゲーム理論やメカニズムデザインは参加者が合理的であることを前提にしている。行動経済学の知見をどう取り入れるか
