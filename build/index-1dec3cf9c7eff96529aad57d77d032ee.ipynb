# Ethereum

## Ethereumとは

**Ethereum** は、任意のプログラム（**スマートコントラクト**）をブロックチェーン上で実行できるプラットフォーム。Vitalik Buterinが2013年にホワイトペーパーを公開し、2015年に稼働した。ネイティブ通貨は **ether（ETH）**。

Bitcoinを「電卓」とすると、Ethereumはプログラムを保持して実行できる「コンピューター」にあたる、という比喩がよく使われる（Ethereumは「ワールドコンピューター」を標榜している）。

## Bitcoinの一般化としてのEthereum

EthereumはBitcoinを2つの方向に一般化したものと捉えられる。

| 一般化 | Bitcoin | Ethereum |
| --- | --- | --- |
| 処理 | 送金（と限定的なScript） | 任意のプログラム（スマートコントラクト） |
| 管理対象 | bitcoinの移転記録（UTXO） | アカウントが持つ状態（state）の遷移 |

例えば「AliceからBobに0.3 ETH送金せよ。ただしこの処理がすでに4回行われていたら、5回目以降はCarolに送金せよ」という処理を考える。これを実現するには、

- 条件分岐を含むプログラムを実行できること
- 「これまでに何回実行されたか」という **状態** を保持できること

が必要になる。BitcoinのScriptはループがなく状態も持てないため、このような処理は書けない。

| 項目 | Bitcoin | Ethereum |
| --- | --- | --- |
| 目的 | 送金ネットワーク | 汎用の分散コンピューター |
| データモデル | UTXOモデル | アカウントモデル（ステートモデル） |
| 実行環境 | Script（非チューリング完全） | EVM（チューリング完全。ただしgasで計算量を制限） |
| コンセンサス | PoW | PoS（2022年まではPoW） |
| ブロック間隔 | 約10分 | 12秒 |
| ハッシュ関数 | SHA-256 | Keccak-256 |
| 発行上限 | 2,100万BTC | 上限なし（手数料の一部を焼却する仕組みあり） |

## ETHの単位

最小単位は **wei**。gasの価格には **gwei** がよく使われる。

| 単位 | wei換算 |
| --- | --- |
| wei | $1$ |
| gwei（shannon） | $10^9$ |
| ether | $10^{18}$ |

ETHはスマートコントラクトを動かすための燃料代（gas代）の支払いに使われることから、「bitcoinはデジタルゴールド、etherはデジタルオイル」と例えられる。PoS移行後は、ネットワークのセキュリティを担保するためのステーク資産としての役割も大きい。

## 主なアップグレード

Ethereumはハードフォークによって仕様を継続的に更新している。PoS移行後は、実行レイヤー（Execution Layer）とコンセンサスレイヤー（Consensus Layer）のアップグレード名を組み合わせた名前が付けられている（例：Shanghai + Capella = Shapella）。

| 時期 | 名称 | 主な内容 |
| --- | --- | --- |
| 2015/07 | Frontier | メインネット稼働 |
| 2016/07 | （DAOフォーク） | The DAO事件の資金を返還するためのハードフォーク。反対派はEthereum Classicとして分岐 |
| 2021/08 | London | EIP-1559（手数料のbase fee導入とburn） |
| 2022/09 | The Merge | PoWからPoSへ移行 |
| 2023/04 | Shapella | ステークしたETHの引き出しが可能に |
| 2024/03 | Dencun | EIP-4844（Proto-Danksharding）。L2向けの安価なデータ領域（blob）を導入 |
| 2025/05 | Pectra | EIP-7702（EOAにコントラクトのコードを委任）、EIP-7251（1バリデータあたりの上限を2048 ETHに引き上げ）、blob数の増加 |
| 2025/12 | Fusaka | PeerDAS（EIP-7594）によるデータ可用性サンプリング、ブロック提案者の事前決定（EIP-7917）など |
