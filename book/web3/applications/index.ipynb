# アプリケーション

## DApps

**DApps（Decentralized Applications, 分散型アプリケーション）** は、中心的な処理にスマートコントラクトを使うアプリケーション。プログラムの実行、実行結果（ステート）の管理、実行履歴（ブロックチェーン）の管理を、見知らぬ誰か（ノード）が担ってくれる。運営企業や政府がその気になっても、記録を改ざん・消去できる構造になっていない。

一方で、普通のWebアプリケーションと比べると明確な弱点がある。

| 弱点 | 理由 |
| --- | --- |
| 遅い | ブロックに取り込まれ、確定するまで待つ必要がある（→[スケーリング](../scaling/index.ipynb)） |
| 計算能力が低く、手数料が高い | 全ノードが同じ計算をやり直すため。EthereumのL1の計算能力はゲームボーイ程度（数MIPS）と言われる |
| 修正しにくい | デプロイしたコードは原則として変更できない |
| 隠せない | ソースコードも実行履歴も誰でも見られる（誰でも検証できることの裏返し） |

これらはトラストレスな合意形成の仕組みから必然的に生じるもの。したがって、

- 運営者を信頼してよいなら、無理にDAppsにする必要はない
- DAppsにする場合も、**ブロックチェーンに載せる部分と載せない部分を見極める** ことが重要

になる。典型的には、見た目や操作感（Webページ、アニメーション）は普通のサーバーで提供し、お金や権利に関わる部分（資産の移転、所有の記録、投票）だけをスマートコントラクトに任せる。

## トークン

多くのDAppsは独自の **トークン** を発行している。

- **ガバナンストークン**：DAppsの仕様変更やパラメーター変更の投票権（→[DAO](./dao.ipynb)）
- **ユーティリティトークン**：サービス内での支払いや利用権
- **LPトークン、預金トークン**：プールに預けた資産の持ち分を表す

発行の動機としては、プロトコル（BitcoinやEthereum）が持つ自律分散的なインセンティブ設計をDAppsでも再現したいという理念と、手っ取り早く資金を集めたいという実態の両面がある。トークンの供給量や配分、価値の源泉の設計は **トークノミクス（tokenomics）** と呼ばれる。

## 主なカテゴリー

| カテゴリー | 内容 | ページ |
| --- | --- | --- |
| DeFi | 分散型金融。取引所、レンディング、ステーブルコインなど | [DeFi](./defi/index.ipynb) |
| DAO | トークン保有者の投票で資金の使い道やルールを決める組織 | [DAO](./dao.ipynb) |
| NFT | 唯一性のあるトークン。アート、会員証、ゲームアイテムなど | [NFT](./nft.ipynb) |
| 予測市場 | 将来の出来事に連動する証券の取引を通じて、確率を集合知で予測する | [予測市場](./prediction_markets.ipynb) |
| エンタープライズ | 企業間のデータ共有基盤としてのコンソーシアム型ブロックチェーン | [エンタープライズブロックチェーン](./enterprise_blockchain.ipynb) |
| RWA | 国債、金、不動産などの現実資産をトークン化したもの | [DeFi](./defi/index.ipynb) |
| DePIN | 通信網やストレージなどの物理インフラを、トークン報酬で分散的に構築する | — |
| GameFi | 暗号資産を稼げるゲーム | — |

2025〜2026年にかけては、ステーブルコイン、RWA、予測市場が大きく伸びている。投機的なブームから、送金・資産保有・運用といった実用への移行が進みつつある。
