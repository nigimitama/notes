# NFT

## NFTとは

**NFT（Non-Fungible Token, 非代替性トークン）** は、1つ1つが固有のIDを持ち、互いに区別されるトークン。ブロックチェーン上で「固有性」と「所有権の移転履歴」を保証するデジタルな証明書として機能する。

BTCやETH、ERC-20トークンは、どの1枚も同じ価値を持つ **代替可能（fungible）** な資産である。お札でいえば、どの千円札も千円の価値なのと同じ。NFTは、絵画の原画や不動産の登記のように、1点ごとに異なるものを表す。

### 技術的な構成

| 要素 | 内容 |
| --- | --- |
| コントラクト | ERC-721（1点ものを1つずつ発行）またはERC-1155（複数種類、各種類を複数枚発行できる）に準拠する（→[トークン規格](../ethereum/token_standards.ipynb)） |
| トークンID | コントラクト内で一意の番号。「コントラクトアドレス + トークンID」でNFTが特定される |
| メタデータ | 名前、説明、画像のURL、属性などのJSON。`tokenURI(tokenId)` で場所を取得する |
| 画像などのデータ | 通常はサイズが大きいためオンチェーンに置かず、IPFS、Arweave、Webサーバーに置く |

```json
{
  "name": "Example #42",
  "description": "...",
  "image": "ipfs://bafybeigdyrzt5sfp7udm7hu76uh7y26nf3efuylqabf3oclgtqy55fbzdi/42.png",
  "attributes": [
    { "trait_type": "Background", "value": "Blue" },
    { "trait_type": "Eyes", "value": "Laser" }
  ]
}
```

### NFTが保証すること・しないこと

NFTがブロックチェーンで保証するのは、**「このコントラクトのこのIDのトークンを、このアドレスが持っている」こと** だけである。

- 画像の著作権や利用権は、NFTの保有とは別の話で、発行者が定める利用規約次第
- 画像がWebサーバーにあれば、サーバーが止まると画像は見られなくなる（IPFSでも誰かがデータを保持し続ける必要がある）
- 誰でも同じ画像を使って別のNFTを発行できる。どのNFTが「本物」かは、発行元のコントラクトアドレスで判断するしかない

データの永続性を高めるため、画像データ自体をコントラクトに保存する **フルオンチェーンNFT**（Nouns, Autoglyphsなど）もある。

## 歴史と市場

| 年 | 出来事 |
| --- | --- |
| 2017 | CryptoPunks（1万体のドット絵のキャラクター、無料配布）、CryptoKitties（猫を交配して新しい猫を生み出すゲーム。Ethereumを混雑させた）。ERC-721が策定される |
| 2020〜2021 | Art Blocks（ジェネラティブアート）、NBA Top Shot、Bored Ape Yacht Club などのPFP（プロフィール画像）コレクション |
| 2021 | Beepleの作品「Everydays: The First 5000 Days」がクリスティーズで約6,900万ドルで落札される。NFTブームのピーク |
| 2022〜2023 | 取引量・利用者数ともに急減。多くのプロジェクトが資金難やコミュニティの離脱で崩壊 |
| 2023〜 | Bitcoin上のNFT（Ordinals）の登場。会員証、チケット、RWAなど実用的な用途へのシフト |

ブームが崩壊した背景には、

- NFTを発行して資金を集めること自体が目的化し、新規購入者の資金で既存の保有者に還元するポンジ的な構造になっていたプロジェクトが多かった
- アニメ化やゲーム化など、大規模な資金が必要な目標を掲げたものの、実現前に資金が尽きた

ことがある。

### マーケットプレイスとロイヤリティ

NFTは OpenSea、Blur、Magic Eden などのマーケットプレイスで売買される。

NFTの特徴として当初期待されたのが、転売のたびに作者に **ロイヤリティ** が支払われる仕組み。ERC-2981でロイヤリティ情報を取得する方法は標準化されたが、**支払いを強制する仕組みではない** ため、手数料競争の中でロイヤリティを任意にするマーケットプレイスが増え、作者の収入源としての機能は弱まった。

## NFTの分類

| カテゴリー | 発行の目的 | 例 | 特徴 |
| --- | --- | --- | --- |
| デジタルアート | 作品の販売・流通の機会をつくる | Beeple, Art Blocks | 作家性や文化的な文脈を活かせるが、需要の波に左右される |
| PFPコレクション | ファンコミュニティの形成とIPの拡張 | CryptoPunks, Bored Ape Yacht Club, Pudgy Penguins | 拡散力があるが、流行に依存し、コミュニティの維持が課題 |
| ゲーム資産 | プレイヤー間の経済を作る | Axie Infinity, STEPN | 経済への参加が活発だが、トークン設計次第で持続性に大きな差が出る |
| 会員証・チケット | 顧客との継続的な関係を可視化する | イベントチケット、ファンクラブ | ユーザー体験に直結するが、期限後の価値の維持が難しい |
| RWA（実物資産） | 資産の流動化・分割所有 | 不動産、トレーディングカード、ワイン | 実需に結びつくが、法規制や実物の保管・監査が課題 |
| SBT（譲渡不可能な証明書） | 資格・履歴・本人性の証明 | 修了証、POAP（イベント参加証明） | 市場価値ではなく信頼性が重要。発行者への信頼に依存する |

### ジェネラティブアート

**ジェネラティブアート** は、アルゴリズムで生成される作品。Art Blocksでは、作品を生成するコードをオンチェーンに保存し、購入（ミント）時のトランザクションハッシュを乱数のシードにして、その場で1点ものの作品が生成される。作者自身も、どんな作品がミントされるかを事前には知らない。

PFPコレクションでも、背景・体・顔などのパーツをランダムに組み合わせて1万体前後を生成し、パーツの出現頻度（レアリティ）が価格に影響することが多い。乱数はブロックの情報から作ると操作されうるため、Chainlink VRFなどが使われる。

### 実用化の事例

近年は、NFTを意識させない形での活用が増えている。

- 動画配信サービスの会員に、毎月デジタルのピンバッジNFTを無料で配る（Disney Pinnacle）
- スポーツチームがレースに合わせてNFTを無料配布し、初めて暗号資産ウォレットを作るファンを大量に獲得する
- メールアドレスだけで作れる埋め込み型のウォレットや、gas代の肩代わり（→[アカウント抽象化](../ethereum/token_standards.ipynb)）で、ユーザーに秘密鍵やgasを意識させない
