# Web3

## Web3とは

**Web3**（Web 3.0）は、ブロックチェーンを基盤として、特定の企業や管理者に依存せずにデータや価値をやりとりできるインターネットを目指す構想の総称。

Webの変遷はしばしば次のように整理される。

| 世代 | 時期の目安 | 特徴 | 代表例 |
| --- | --- | --- | --- |
| Web1 | 1990年代〜 | 読むだけ（Read）。静的なページを一方向に配信 | 個人ホームページ、ポータルサイト |
| Web2 | 2000年代中盤〜 | 読み書きできる（Read / Write）。ユーザー生成コンテンツ。ただしデータとアカウントはプラットフォーム企業が管理 | SNS、検索エンジン、EC |
| Web3 | 2010年代後半〜 | 読み書きに加えて所有できる（Read / Write / Own）。資産・データ・ルールをブロックチェーン上に置き、プラットフォームへの信頼を不要にする | 暗号資産、DeFi、NFT、DAO |

「Web3」という言葉自体はEthereumの共同創設者Gavin Woodが2014年頃に使い始めたもので、明確な技術的定義があるわけではない。マーケティング用語として使われる場面も多い点には注意。

## なぜブロックチェーンなのか

### 信頼できる第三者への依存

世の中のシステムのほとんどは、集権的な管理者（運営会社）がいることを前提にしている。

- 決済：銀行が不正な送金や違法な口座を監視する
- 検索：検索エンジン企業が違法サイトを結果から除外する
- 保険：保険会社が被保険者の被害の真偽を調査する

しかし管理者自体が不正を働く可能性もある（銀行が預金を流用する、検索結果を自社に都合よく操作する、など）。それを防ぐために監督機関を置いても、今度は「その監督機関を誰が監督するのか」という問題が残る。これは古くから **「誰が見張りを見張るのか？」（Quis custodiet ipsos custodes?）** と呼ばれる問題。

### トラストレス

ブロックチェーンの基本的な価値は、**信頼できる第三者（trusted third party）に頼らずに、参加者全員が同じ記録に合意できる**こと。Bitcoinの論文（[Nakamoto, 2008](https://bitcoin.org/bitcoin.pdf)）の冒頭にも「信用ではなく暗号学的証明に基づく電子決済システム」を目指すと書かれている。

この性質は **トラストレス（trustless）** と呼ばれる。「システムを機能させるうえで、管理者や個々の参加者を信用する必要がない」という意味であり、

- 暗号技術（ハッシュ関数・デジタル署名）によって記録の改ざんやなりすましを検知可能にし
- 経済的インセンティブ（報酬とコスト）によって、不正をするより正直に振る舞うほうが得になるようにする

ことで実現している。不正が割に合わない仕組みを作れれば、そもそも見張りは要らない、という発想である。

:::{note}
オープンソースであることとトラストレスであることは別物。ソースコードが公開されていても、運営者が自由にデータを書き換えられるならトラストレスではない。
:::

### ブロックチェーンが不要な場面

逆に言えば、信頼できる第三者に任せて問題ない場面では、無理にブロックチェーンを使う必要はない。ブロックチェーンは通常のデータベースと比べて

- 処理性能が低い（スループットが小さく、確定まで時間がかかる）
- 手数料がかかる
- データを後から修正・削除できない
- 外部世界の情報を直接扱えない（オラクル問題）

といった制約がある。「管理者を信用できない／したくない複数の主体が、共通の記録を持ちたい」という状況でこそ価値がある。

## 歴史

| 年 | 出来事 |
| --- | --- |
| 1991 | Haber & Stornetta がハッシュの連鎖によるタイムスタンプ方式を提案（ブロックチェーンの原型） |
| 2008 | Satoshi Nakamoto がBitcoinの論文を公開 |
| 2009 | Bitcoinのネットワークが稼働（Genesisブロック） |
| 2013 | Vitalik Buterin がEthereumのホワイトペーパーを公開 |
| 2015 | Ethereumのメインネットが稼働 |
| 2016 | The DAO事件（スマートコントラクトの脆弱性による資金流出、Ethereumのハードフォーク） |
| 2017 | ICOブーム、CryptoKitties、BitcoinでSegWit導入 |
| 2020 | DeFi Summer（Uniswap、Compound等の急成長） |
| 2021 | NFTブーム、BitcoinでTaproot導入 |
| 2022 | Ethereumが The Merge でPoWからPoSへ移行。Terra/LUNAやFTXの破綻 |
| 2024 | EthereumのDencunアップグレード（EIP-4844）でL2の手数料が大幅低下。米国でBitcoin現物ETF承認 |
