# Berry(1994)による需要推定

Steven Berry (1994)「Estimating Discrete-Choice Models of Product Differentiation」(RAND Journal of Economics) は、差別化財市場における需要推定の内生性問題を解決する手法を提示した論文です。BLP (Berry, Levinsohn, Pakes 1995) の基礎となった

Berry (1994) の解決策:シェア逆転法(Inversion) - 個人レベルのデータがなくても、市場シェアデータから観察不能な財の平均効用を「逆転」して取り出せることを示した。

→ 離散選択モデル（ロジットモデル）を整理して線形回帰モデルを導出し、 **線形回帰や操作変数法で需要推定ができる** ことを示した。

消費者$i$がマーケット$t$（地理や時点の差異などを示す）において製品$j$を購入した場合の効用$U_{i j t}$は以下のように与えられる

$$
U_{i j t}= \begin{cases}\beta^0+\sum_{k=1}^K \beta^k x_{j t}^k-\alpha p_{j t}+\xi_{j t}+\epsilon_{i j t}, & j=1, \ldots, J_t \\ \epsilon_{i 0 t}, & j=0\end{cases}
$$

ロジットモデルでは製品$j$の購入確率は次のように表される

$$
\operatorname{Pr}(j \text { を購入 })
=\frac{\exp \left(\beta^0+\sum_{k=1}^K \beta^k x_{j t}^k-\alpha p_{j t}+\xi_{j t}\right)}{1+\sum_{l=1}^{J_t} \exp \left(\beta^0+\sum_{k=1}^K \beta^k x_{l t}^k-\alpha p_{l t}+\xi_{l t}\right)}
$$

この確率をすべての消費者について集約することで、マーケット全体の製品需要が得られる。  
製品$j$への需要$q_{j t}(p_{1 t}, \ldots, p_{J_t t})$は、潜在的な消費者の数が$M_t$のとき以下のように定義する。

$$
q_{j t} = q_{j t}(p_{1 t}, \ldots, p_{J_t t})=M_t \times \operatorname{Pr}(j \text { を購入})
$$



:::{margin}

「潜在的な」消費者の数を使うのは「製品を買わない（アウトサイドグッズ）」という選択肢もあるため。

:::



マーケット$t$, 製品$j$マーケットシェア$s_{jt}$を次のように定義する

$$
s_{j t}:= \begin{cases}\frac{q_{j t}}{M_t}, & j=1, \ldots, J_t \\ \frac{M_t-\sum_{j=1}^{J_t} q_{j t}}{M_t}, & j=0\end{cases}
$$

$q_{j t} = M_t \times \operatorname{Pr}(j \text { を購入})$ なので、$\frac{q_{j t}}{M_t} = \operatorname{Pr}(j \text { を購入})$ であるから、マーケットシェアに代入すると

$$
s_{j t}= \begin{cases}
\frac{\exp \left(\delta_{j t}\right)}{1+\sum_{l=1}^{J_t} \exp \left(\delta_{l t}\right)}, & j=1, \ldots, J_t \\
\frac{1}{1+\sum_{l=1}^{J_t} \exp \left(\delta_{l t}\right)}, & j=0
\end{cases}
$$

ここで $\delta_{j t}$ は消費者間で共通な効用の要素で、 **平均効用(mean utility)** と呼ばれる。なおアウトサイド・グッズの平均効用$\delta_{0t}$はゼロに基準化している（$\delta_{0t}=0$）とする。

$$
\delta_{j t}=\beta^0+\sum_{k=1}^K \beta^k x_{j t}^k-\alpha p_{j t}+\xi_{j t}
$$

$s_{jt}$を$s_{0t}$で割り自然対数をとると、以下の式が得られる。

:::{admonition} Berry inversion

$$
\log \left(\frac{s_{j t}}{s_{0 t}}\right)=\delta_{j t}
$$

:::

つまり、 **本来観察できない平均効用 $\delta_{jt}$ が、観測可能なマーケット・シェアの比率（の対数）と等しい** ことを示している。


:::{margin}
マーケットシェアは平均効用によって決定されると想定しているが、観測したマーケットシェアから平均効用に逆変換できるということから **Berry inversion（Berryの逆変換）** と呼ばれる
:::

平均効用を全部書くと、

:::{card}


$$
\log \left(\frac{s_{j t}}{s_{0 t}}\right)
= \beta^0+\sum_{k=1}^K \beta^k x_{j t}^k-\alpha p_{j t}+\xi_{j t}
$$

:::

→ シェアの比率の対数値を価格と製品属性に回帰する線形回帰モデルとして書くことができる

## 内生性問題

しかしながら、単純に線形回帰でパラメータを推定すると、価格と誤差項が相関するため欠落変数バイアスによりパラメータを過小評価する。誤差項に含まれる、未観測の属性（例えばブランドイメージ）が価格と相関するためである。

内生性問題への代表的な対処法は操作変数法である。  
操作変数とは、価格とは相関するが誤差項とは相関しないような変数で、次のようなものが用いられる。

### 操作変数1. 生産の費用情報

価格の操作変数として理想的なのは製品の生産コストに関する情報（コスト・シフター cost shifter）である。  
このアイデアは教科書的な同質財の需要・供給曲線の推定で用いられる。　　
しかし差別化財の需要関数に直接適用するのは難しいケースが多い。市場のすべての製品についてのコスト情報を得られることが稀であるため。


### 操作変数2. BLP操作変数・差別化操作変数

分析でよく利用される変数は「市場における製品間の競争状況」を捉える変数。BLP操作変数と、それを改善した差別化操作変数が存在する。

#### BLP操作変数

1. 同じ企業が生産している他の製品属性の和

$$
z_{j t}^{B L P, \text { other }}=\sum_{k \in J_{f t}, k \neq j} x_{k t}
$$

と、

2. 他の企業が生産している製品属性の和

$$
z_{j t}^{B L P, \text { rival }}=\sum_{k \notin J_{f t}} x_{k t}
$$

で定義される。ここで$J_{ft}$は企業$f$がマーケット$t$において販売している製品の集合。
